# Phase 3: Data Exploration
In this notebook, we explore the Student Mood/Depression dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../data/student_dataset.csv')
```

## 2. Visualizations

```python
# Class distribution
plt.figure(figsize=(6, 4))
sns.countplot(x='Depression', data=df, palette='viridis')
plt.title('Class Distribution (Depression)')
plt.show()
```

```python
# Histograms
df.hist(bins=20, figsize=(10, 8), color='skyblue')
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(8, 6))
# Compute correlation only for numeric columns
numeric_cols = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Boxplots for numerical features
plt.figure(figsize=(10, 6))
sns.boxplot(data=numeric_cols, orient='h', palette='Set2')
plt.title('Boxplots of Numerical Features')
plt.show()
```
